```python
from hf_dataset_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer
```

```python
cargador = HFDatasetLoader("scikit-learn/iris")

df = cargador.obtener_dataframe()

print("Dataset cargado correctamente")
print(df.shape)
display(df.head())
```

```python
# Transformación: convertir los nombres de las columnas a minúsculas

df_minusculas = df.copy()
df_minusculas.columns = [columna.lower() for columna in df_minusculas.columns]

print("Columnas después de la transformación:")
print(df_minusculas.columns.tolist())

display(df_minusculas.head())
```

```python
resumidor = DatasetSummarizer(df_minusculas)

resumidor.generar_resumen("resumen_dataset.txt")
```

```python
with open("resumen_dataset.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
```
